```python
from sklearn.cluster import KMeans

# Each row is [height, weight]
people = [
    [150, 50],
    [155, 52],
    [160, 55],
    [180, 80],
    [185, 85],
    [190, 90]
]

# Create a k-means model
model = KMeans(n_clusters=2, random_state=42, n_init=10)

# Find the 2 groups
model.fit(people)

# Show which group each person belongs to
print(model.labels_)
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# [amount spent, number of visits]
customers = [
    [20, 1],
    [25, 2],
    [30, 2],
    [35, 3],
    [100, 8],
    [110, 9],
    [120, 10],
    [130, 11]
]

# Create the k-means model
model = KMeans(n_clusters=2, random_state=42, n_init=10)

# Find the groups
model.fit(customers)

# Get the group for each customer
groups = model.labels_

# Separate the x and y values
spending = [customer[0] for customer in customers]
visits = [customer[1] for customer in customers]

# Make a scatter plot
plt.scatter(spending, visits, c=groups)

plt.xlabel("Amount spent ($)")
plt.ylabel("Number of visits")
plt.title("Customer Groups")

plt.show()
```

```python
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt

# Create some fake data
X, y = make_blobs(
    n_samples = 1_000,
    centers=5,
    random_state=42
)

# Plot the data
plt.scatter(X[:, 0], X[:, 1])
plt.show()
```

```python
# Create k-means
model = KMeans(n_clusters=5, random_state=42, n_init=10)

# Find the clusters
model.fit(X)

# Get the predicted groups
clusters = model.labels_

# Plot the clusters
plt.scatter(X[:, 0], X[:, 1], c=clusters)

plt.show()
```
